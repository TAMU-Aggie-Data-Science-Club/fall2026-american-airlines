```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#Loading the data sets
df_pax = pd.read_csv('../../data/AA_AIRPORT_INFO.csv')
df_airports = pd.read_csv('../../data/AA_AIRPORT_INFO.csv')

#displaying the first few rows of the data sets
print("PAX Data Head:")
display(df_pax.head())

print("Airport Data Head:")
display(df_airports.head())
```

```python
#Checking if there is missing info for airports
print("PAX Data Info")
print(df_pax.info())

print("PAX Data Info")
print(df_pax.isnull().sum())
```

```python
#Checking the basic stats
display(df_pax.describe())
```

```python
#PLotting Correlation Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df_pax.select_dtypes(include=['number']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap - PAX Data')
plt.show()
```

```python
#Fill the missing values with place holder unknown
df_pax['STATE_PROVNC_CD'] = df_pax['STATE_PROVNC_CD'].fillna('Unknown')
df_pax['CNTRY_CD'] = df_pax['CNTRY_CD'].fillna('Unknown')

#checking if zero missing vales are still there
print(df_pax.isnull().sum())
```

```python
#saving processed data for modeling phase
df_pax.to_csv('../../data/cleaned_pax_data.csv', index=False)
```
